# Sección 2 — Inventario NASA OSDR

Objetivo:
Construir un inventario reproducible de estudios de Arabidopsis thaliana disponibles en NASA OSDR/GeneLab.

Fuente:
NASA OSDR Biological Data API.

```python
import pandas as pd
import requests
from io import StringIO

BASE_URL = "https://visualization.osdr.nasa.gov/biodata/api/v2/query/metadata/"

params = {
    "study.characteristics.organism": "Arabidopsis thaliana",
    "id.accession": ""
}

response = requests.get(BASE_URL, params=params, timeout=60)

print("Status code:", response.status_code)
print("URL:", response.url)
```

```python
study_inventory = (
    df.groupby("id.accession", as_index=False)
      .agg(
          organism=("study.characteristics.organism", "first"),
          n_samples=("id.sample name", "nunique"),
          n_assays=("id.assay name", "nunique"),
          assay_names=("id.assay name", lambda x: " | ".join(sorted(set(x))))
      )
)

study_inventory.head(10)
```

```python
def classify_omics(assay_name):
    assay = assay_name.lower()

    if "rna-seq" in assay or "rna sequencing" in assay:
        return "RNA-seq"

    elif "microarray" in assay:
        return "Microarray"

    elif "bisulfite" in assay or "methyl" in assay:
        return "Epigenomics"

    elif "proteom" in assay:
        return "Proteomics"

    elif "metabolom" in assay:
        return "Metabolomics"

    elif "small rna" in assay or "mirna" in assay:
        return "Small RNA-seq"

    else:
        return "Other"
```

```python
study_inventory["omics_type"] = study_inventory["assay_names"].apply(classify_omics)

study_inventory[
    ["id.accession", "omics_type", "n_samples", "n_assays", "assay_names"]
].head(20)
```

```python
def contains_any(text, terms):
    text = str(text).lower()
    return any(term.lower() in text for term in terms)


study_inventory["has_rna_seq"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "rna-sequencing",
            "rna sequencing",
            "rna-seq",
            "transcription-profiling_rna"
        ]
    )
)

study_inventory["has_small_rna"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "small rna",
            "small-rna",
            "mirna",
            "microRNA"
        ]
    )
)

study_inventory["has_microarray"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "microarray",
            "dna-microarray"
        ]
    )
)

study_inventory["has_methylation"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "methylation",
            "bisulfite"
        ]
    )
)

study_inventory["has_proteomics"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "protein-expression",
            "proteom",
            "mass-spectrometry"
        ]
    )
)

study_inventory["has_metabolomics"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "metabolom"
        ]
    )
)

study_inventory["has_glycomics"] = study_inventory["assay_names"].apply(
    lambda x: contains_any(
        x,
        [
            "glycomic",
            "glycome",
            "elisa"
        ]
    )
)
```

```python
def summarize_study(osd_id):
    url = "https://visualization.osdr.nasa.gov/biodata/api/v2/query/metadata/"

    params = {
        "id.accession": osd_id,
        "study.characteristics": "",
        "study.factor value": "",
        "study.parameter value": ""
    }

    r = requests.get(url, params=params, timeout=60)
    r.raise_for_status()

    df_study = pd.read_csv(StringIO(r.text))

    def unique_values(column):
        if column not in df_study.columns:
            return None

        values = (
            df_study[column]
            .dropna()
            .astype(str)
            .unique()
            .tolist()
        )

        return " | ".join(values) if values else None

    return {
        "OSD_ID": osd_id,
        "n_samples": df_study["id.sample name"].nunique()
            if "id.sample name" in df_study.columns else None,

        "spaceflight": unique_values("study.factor value.spaceflight"),
        "age": unique_values("study.factor value.age"),
        "cultivar": unique_values("study.factor value.cultivar"),
        "genotype": unique_values("study.characteristics.genotype"),
        "organ": unique_values("study.characteristics.organism part"),
        "developmental_stage": unique_values(
            "study.characteristics.developmental stage"
        ),

        "growth_environment": unique_values(
            "study.parameter value.growth environment"
        ),

        "hardware": unique_values(
            "study.parameter value.hardware"
        ),

        "light_cycle": unique_values(
            "study.parameter value.light cycle"
        ),

        "light_intensity": unique_values(
            "study.parameter value.light intensity"
        ),

        "light_quality": unique_values(
            "study.parameter value.light quality"
        ),

        "light_source": unique_values(
            "study.parameter value.light source"
        ),

        "temperature": unique_values(
            "study.parameter value.mean growth temperature"
        )
    }
```

```python
from time import sleep

all_summaries = []
errors = []

osd_ids = study_inventory["id.accession"].tolist()

for i, osd_id in enumerate(osd_ids, start=1):
    print(f"[{i}/{len(osd_ids)}] Procesando {osd_id}...")

    try:
        summary = summarize_study(osd_id)
        all_summaries.append(summary)

    except Exception as e:
        errors.append({
            "OSD_ID": osd_id,
            "error": str(e)
        })

        print(f"  Error en {osd_id}: {e}")

    sleep(0.2)
```

```python
metadata_summary = pd.DataFrame(all_summaries)

print("Estudios procesados:", len(metadata_summary))
print("Errores:", len(errors))

metadata_summary.head()
```

```python
pd.DataFrame(errors)
```
